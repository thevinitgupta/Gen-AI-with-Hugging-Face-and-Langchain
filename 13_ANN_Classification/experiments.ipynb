```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
# Load the Dataset
data = pd.read_csv("Churn_Modelling.csv")
data.head()
```

```python
## Preprocess Data
### Remove Irrelevant Columns
data = data.drop(["RowNumber","CustomerId", "Surname"], axis=1)
data.head()
```

```python
# Encode Categorical Values
label_encoder_gender = LabelEncoder()
data['Gender'] = LabelEncoder.fit_transform(label_encoder_gender,data['Gender'])
data.head
```

```python
# OneHotEncoding for Geography Col
from sklearn.preprocessing import OneHotEncoder
geo_onehot_encoder = OneHotEncoder(sparse_output=False)
encoded_geography = geo_onehot_encoder.fit_transform(data[['Geography']])
encoded_geography
```

```python
geo_onehot_encoder.get_feature_names_out(['Geography'])
```

```python
geo_features_df = pd.DataFrame(encoded_geography,columns=geo_onehot_encoder.get_feature_names_out(['Geography']))
geo_features_df
```

```python
## Combine One-Hot Encoded columns with the original data
data = pd.concat([data.drop('Geography', axis=1), geo_features_df], axis=1)
data
```

```python
## save encoders
with open('label_encoder_gender.pkl', 'wb') as file:
    pickle.dump(label_encoder_gender, file)

with open('onehot_encoder_geo.pkl', 'wb') as file:
    pickle.dump(geo_onehot_encoder, file)
```

#### What `train_test_split` does conceptually ￼

You usually have:

- `X`: input features (e.g., columns like age, salary, etc., or pixel values if images)

- `y`: target labels (e.g., 0/1 for classification, or a numeric value for regression)

When training a model, you don’t train and evaluate on the same data, or you’ll overestimate performance. So you split your data into:

- Training set: used to fit the model’s parameters.

- Test set: held out and used only to evaluate how well the trained model generalizes to unseen data.

`train_test_split` automates this splitting while keeping `X` and `y` aligned (same order of samples).

Left-hand side: the four outputs ￼￼

This means `train_test_split` returns four arrays (or DataFrames):

1. `X_train`: features for training

2. `X_test`: features for testing

3. `y_train`: labels corresponding to `X_train`

4. `y_test`: labels corresponding to `X_test`

The important point: the i-th row of `X_train` corresponds to the i-th value of `y_train`, and similarly for `X_test`/`y_test`.

Right-hand side: the inputs and parameters ￼

1) `X, y` ￼

- `X` is your full feature set (often a NumPy array or pandas DataFrame of shape `(n_samples, n_features)`).

- `y` is your full target vector (array/Series of length `n_samples`).

`train_test_split` takes these and shuffles/splits them in the same way so the pairing doesn’t break.

2) `test_size=0.2` ￼

This controls how much of the data goes into the test set.

- `test_size=0.2` means 20% of the total samples will be used as test data.

- The remaining 80% will be used as training data.

Example: If you have 1000 samples:

- `X_train` / `y_train`: 800 samples

- `X_test` / `y_test`: 200 samples

If you had set `test_size=0.3`, then 30% would be test data, 70% training, etc.

3) `random_state=43` ￼

`train_test_split` shuffles rows before splitting, so you don’t get a biased split (e.g., all early rows in train, all later rows in test).

`random_state` is the seed for the random number generator used to perform this shuffle and split.

- With the same `random_state` (here 43), the split will be reproducible.

- Every time you run this line with the same `X`, `y`, `test_size` and `random_state=43`, you will get the exact same `X_train`, `X_test`, `y_train`, `y_test`.

- If you change `random_state` (or omit it), you’ll likely get a different split each run.

This is important for:

- Comparing models fairly.

- Debugging (so that results don’t randomly change between runs).

- Reproducing experiments.

Rough internal steps of what happens ￼

Under the hood, `train_test_split(X, y, test_size=0.2, random_state=43)` roughly does:

1. Compute how many samples should be in the test set: `n_test = int(0.2 * n_samples)`.

2. Initialize a random number generator with seed `43`.

3. Shuffle the indexes `0, 1, 2, ..., n_samples-1` in a random order.

4. Take the first `n_test` indexes as test indexes, and the rest as train indexes.

5. Use those indexes to slice `X` and `y`:

 ▫ `X_train = X[train_indexes]`

 ▫ `X_test = X[test_indexes]`

 ▫ `y_train = y[train_indexes]`

 ▫ `y_test = y[test_indexes]`

Because the same shuffled indices are applied to both `X` and `y`, they stay in sync.

```python
import datetime
```

```python
import tensorflow as tf
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
```

```python
(X_train.shape[1],)
```

```python
# Sequential Model creation
seq_model = Sequential([
    Dense(64,activation="relu", input_shape=(X_train.shape[1],)), # HL1 -> Connected with Input layer so shape defined
    Dense(32,activation="relu"), # HL2
    Dense(1,activation="sigmoid") # output layer -> Binary(0/1)
])
```

```python
seq_model.summary()
```

```python
# optimzers and loss fucntions init
opt = tf.keras.optimizers.Adam(learning_rate=0.002)
loss = tf.keras.losses.BinaryCrossentropy()
```

```python
# model compilation
seq_model.compile(loss=loss, optimizer=opt,metrics=['accuracy'])
```

```python
# setup tensorboard
log_dir = "logs/fit" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tf_callback = TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
# Setup Early Stopping
early_stopping_cb = EarlyStopping(monitor="val_loss",patience=5,restore_best_weights=True)
```
